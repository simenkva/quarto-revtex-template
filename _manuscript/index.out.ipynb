## Workflow

This section documents how the manuscript is built and should be removed before submission. The source is the Quarto file `index.qmd`, and `_quarto.yml` configures a Quarto *manuscript* project that renders HTML, Word, JATS and a RevTeX PDF from that single source.

### Building

Render all formats, or only the PDF, with

``` bash
quarto render
quarto render index.qmd --to pdf
```

Output is written to `_manuscript/`. Because `freeze: true` is set, a project-wide render reuses stored code output; rendering `index.qmd` explicitly re-executes the Python chunks. Use `quarto preview` for live reloading while writing.

### RevTeX setup

The PDF uses `documentclass: revtex4-2` with the options `aps, pra, reprint, superscriptaddress`; change `pra` to target another journal, or `reprint` to `preprint` for a single-column draft. The folder `_revtex/` holds the glue between Quarto and RevTeX. The file `before-body.tex` emits `\title`, `\author`, `\affiliation`, `\email` and the abstract after `\begin{document}`, as RevTeX requires. The file `preamble.tex` resolves a `longtable*` clash with the `caption` package. The generated LaTeX is kept in `_manuscript/_tex/` for journal submission.

### Fonts

Text and mathematics are set in STIX Two in both the PDF and the HTML output. The font files are in `_fonts/`, and they are switched on by `stix: true` in `_quarto.yml`; set it to `false` to use the default fonts. The filter `_fonts/stix.lua` loads the fonts in the PDF preamble and, for HTML, tells MathJax 4 to use its STIX Two math font.

### Plots

The first code chunk in this file loads the local package `revtexplot`, which styles matplotlib figures to match the PDF. It is switched on by `plot-style: true` in `_quarto.yml`; set it to `false` for matplotlib’s defaults. Figures are one column wide by default and use 8 pt text, the size of `\footnotesize`, in STIX Two (or Computer Modern with `stix: false`). They are saved at exactly their set size, so the text in them matches the captions. The package provides these sizes:

- `rp.COLUMN`: one column, 3.40 × 2.10 in (the default)
- `rp.WIDE`: both columns, 7.06 × 2.35 in
- `rp.figsize(width, aspect)`: `width` is `"column"`, `"wide"` or inches, and `aspect` is height / width

For example, `plt.subplots(figsize=rp.figsize("column", 1))` gives a square one-column figure.

### Writing

Authors, affiliations, e-mail addresses and the abstract are set in the YAML header. Sections, equations and figures get labels with the prefixes `sec-`, `eq-` and `fig-`, and are referenced as `@sec-methods`, `@eq-gaussian` or `@fig-example`. A displayed equation is labelled by appending `{#eq-name}` after its closing `$$`. Citations use keys from `references.bib`, e.g. `[@knuth1984]`, and are formatted by BibTeX with the `apsrev4-2` style. Figures are produced by Python chunks whose `label` and `fig-cap` options set the reference and caption; see <a href="#sec-results" class="quarto-xref">Section 4</a> for an example.

### Wide figures and equations

In the two-column `reprint` layout, a figure spans both columns when it uses RevTeX’s `figure*` environment. For a Python chunk, add the options

``` python
#| fig-env: "figure*"
#| fig-pos: "t"
```

and create the figure with `plt.subplots(figsize=rp.WIDE)` (see <a href="#sec-plots" class="quarto-xref">Section 1.4</a>). For an image file, write

``` markdown
![Caption.](figures/wide.pdf){#fig-wide
  fig-env="figure*" fig-pos="t"}
```

Wide figures float to the top of a following page. Always set `fig-pos: "t"`: Quarto places chunk figures with `[H]` by default, which makes a `figure*` vanish from the PDF. Do not use `column: page`; it loads the `sidenotes` package, which stops every `figure*` from spanning both columns.

A long equation is set across both columns by wrapping it in RevTeX’s `widetext` environment with raw LaTeX blocks:

```` markdown
```{=latex}
\begin{widetext}
```
$$
E = \sum_n a_n x^n + \dots
$$ {#eq-wide}
```{=latex}
\end{widetext}
```
````

Labels and cross-references work as usual. The HTML output ignores `fig-env`, `fig-pos` and the raw LaTeX, so both render normally there.

## Methods

### Inline and displayed mathematics

Inline mathematics is written between single dollar signs, as in $e^{i\pi} + 1 = 0$. Displayed equations are written between double dollar signs and numbered when given a label, as for the Gaussian integral <span id="eq-gaussian">
$$\int_{-\infty}^{\infty} e^{-a x^2}\, dx = \sqrt{\frac{\pi}{a}},
\qquad a > 0.
 \qquad(1)$$</span>

### Multi-line equations

Long expressions can be split with an `aligned` environment inside a single labelled equation, <span id="eq-taylor">
$$\begin{aligned}
f(x) &= \sum_{n=0}^{\infty} \frac{f^{(n)}(x_0)}{n!} (x - x_0)^n \\
     &= f(x_0) + f'(x_0)(x - x_0) + \mathcal{O}\big((x - x_0)^2\big).
\end{aligned}
 \qquad(2)$$</span>

### Cross-references

## Results

<a href="#fig-example" class="quarto-xref">Figure 1</a> is generated by the Python code chunk embedded in the source file. It shows the damped oscillation <span id="eq-damped">
$$f(t) = e^{-\gamma t} \cos(\omega t).
 \qquad(3)$$</span> for three values of the damping rate $\gamma$.

`````
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 4, 400)
omega = 2 * np.pi

fig, ax = plt.subplots()
for gamma in (0.25, 0.5, 1.0):
    ax.plot(t, np.exp(-gamma * t) * np.cos(omega * t), lw=1,
            label=rf"$\gamma = {gamma}$")
ax.set_xlabel("$t$")
ax.set_ylabel("$f(t)$")
ax.legend()
plt.show()
`````

<a href="#fig-wide" class="quarto-xref">Figure 2</a> spans both columns and shows each damping rate in its own panel.

`````
fig, axes = plt.subplots(1, 3, figsize=rp.WIDE, sharey=True)
for ax, gamma in zip(axes, (0.25, 0.5, 1.0)):
    ax.plot(t, np.exp(-gamma * t) * np.cos(omega * t))
    ax.plot(t, np.exp(-gamma * t), "k--", lw=0.6)
    ax.plot(t, -np.exp(-gamma * t), "k--", lw=0.6)
    ax.set_title(rf"$\gamma = {gamma}$")
    ax.set_xlabel("$t$")
axes[0].set_ylabel("$f(t)$")
plt.show()
`````
